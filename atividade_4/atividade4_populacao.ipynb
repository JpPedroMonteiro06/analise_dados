# Atividade 4 — População 2010 x 2022 (Censo IBGE)

1. Ler a tabela `CD2022_Populacao_2010_Compatibilizada_20231222.xlsx` (está nesta pasta)
2. Agregar a população por estado
3. Ordenar pelos estados que mais cresceram entre 2010 e 2022 e salvar em CSV
4. Fazer o mesmo por município e salvar em CSV

```python
# python -m pip install pandas openpyxl
import pandas as pd
```

## Ler a tabela

```python
# O cabeçalho está na 3ª linha (skiprows=2) e a primeira coluna é vazia
df = pd.read_excel("CD2022_Populacao_2010_Compatibilizada_20231222.xlsx", skiprows=2)
df = df.drop(columns=df.columns[0])
df.columns = ["UF", "COD_UF", "COD_MUNIC", "MUNICIPIO", "2010_sinopse", "2010", "2022"]

# remove as linhas de nota/fonte do final
df = df.dropna(subset=["UF", "2022"])
for c in ["2010_sinopse", "2010", "2022"]:
    df[c] = pd.to_numeric(df[c]).astype("int64")

print(df.shape)
df.head()
```

## População agregada por estado

Uso a coluna **População 2010 compatibilizada** (limites de 2022), que é a que o IBGE indica para comparar com 2022.

```python
pop_estado = (
    df.groupby("UF")[["2010", "2022"]]
    .sum()
    .reset_index()
)
pop_estado["crescimento"] = pop_estado["2022"] - pop_estado["2010"]
pop_estado["crescimento_%"] = (pop_estado["crescimento"] / pop_estado["2010"] * 100).round(2)

pop_estado = pop_estado.sort_values("crescimento", ascending=False).reset_index(drop=True)
pop_estado
```

```python
pop_estado.to_csv("pop_estado_crescimento.csv", sep=";", index=False)
print("Salvo: pop_estado_crescimento.csv")
```

## Por município

```python
pop_municipio = df[["UF", "MUNICIPIO", "2010", "2022"]].copy()
pop_municipio["crescimento"] = pop_municipio["2022"] - pop_municipio["2010"]
pop_municipio["crescimento_%"] = (pop_municipio["crescimento"] / pop_municipio["2010"] * 100).round(2)

pop_municipio = pop_municipio.sort_values("crescimento", ascending=False).reset_index(drop=True)
pop_municipio.head(15)
```

```python
print("Municípios que mais perderam população:")
pop_municipio.tail(10)
```

```python
pop_municipio.to_csv("pop_municipio_crescimento.csv", sep=";", index=False)
print("Salvo: pop_municipio_crescimento.csv")
```
